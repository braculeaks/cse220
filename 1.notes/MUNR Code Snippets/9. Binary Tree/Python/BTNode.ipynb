## **Node class for Binary Tree**

```python
class BTNode:

  def __init__(self, elem):
    self.elem = elem
    self.left = None
    self.right = None
```

## **Manually linking Tree nodes**

```python
root = BTNode("A")
n1 = BTNode("B")
n2 = BTNode("C")
n3 = BTNode("D")
n4 = BTNode("E")
n5 = BTNode("F")
n6 = BTNode("G")

root.left = n1
root.right = n2

n1.left = n3
n1.right = n4

n2.left = n5
n2.right = n6
```

## **Preorder Traversal**

```python
def preorder(node):
  if node == None:
    return None

  print(node.elem)
  preorder(node.left)
  preorder(node.right)
```

```python
print("The preorder sequence is: ")
preorder(root)
```

## **Inorder Traversal**

```python
def inorder(node):
  if node == None:
    return None

  inorder(node.left)
  print(node.elem, end = " ")
  inorder(node.right)
```

```python
print("The inorder sequence is: ")
inorder(root)
```

## **Postorder Traversal**

```python
def postorder(node):
  if node == None:
    return None

  postorder(node.left)
  postorder(node.right)
  print(node.elem, end = " ")
```

```python
print("The postorder sequence is: ")
postorder(root)
```

## **Creating Binary Tree nodes using Recursion**

```python
def tree_construction(arr, i = 1):
  if i >= len(arr) or arr[i] == None:
    return None
  else:
    p = BTNode(arr[i])
    p.left = tree_construction(arr, 2*i)
    p.right = tree_construction(arr, 2*i + 1)
    return p

newRoot = tree_construction([None, 'A', 'B', 'C', 'D', 'E', 'F', 'G'])
```

```python
print("The inorder sequence is: ")
inorder(newRoot)
```

## **Finding Count of all nodes of the Tree**

```python
def findNoOfNodes(node):
    if node == None:
        return 0
    return 1 + findNoOfNodes(node.left) + findNoOfNodes(node.right)
```

```python
numberOfNodes = findNoOfNodes(root)
print(numberOfNodes)
```

## **Finding Height of a node of the Tree**

```python
def getHeight(node):
  if node == None:
    return -1
  return 1 + max(getHeight(node.left), getHeight(node.right))
```

```python
height = getHeight(n2)
print(f"The height of the node is: {height}")
```

## **Finding Depth of a node of the Tree**

```python
def getDepth(node, target, depth):
    if node == None:
        return -1
    if node.elem == target.elem:
        return depth
    leftDepth = getDepth(node.left, target, depth + 1)
    if leftDepth != -1:
        return leftDepth
    rightDepth = getDepth(node.right, target, depth + 1)
    return rightDepth
```

```python
print(f"Depth of node is: {getDepth(root, n5, 0)}")
```

## **Finding maximum in each level**

```python
def findLevelMax(node, level = 0, level_max = None):
    if node == None:
        return
    if level_max == None:
        level_max = []
    if len(level_max) <= level:
        level_max.append(node.elem)
    else:
        level_max[level] = max(level_max[level], node.elem)
    findLevelMax(node.left, level + 1, level_max)
    findLevelMax(node.right, level + 1, level_max)

    return level_max
```

```python
maximumLevel = findLevelMax(root)
print(maximumLevel)
```

```python
def isValid(node, key, s):
    if node == None:
        return False
    s += node.elem
    if s == key and node.right == None and node.left == None:
        return True
    return isValid(node.left, key, s) or isValid(node.right, key, s)
```

```python
def build_key(root, level):
  if root == None:
    return ""
  result = ""
  result += build_key(root.right, level + 1)
  result += build_key(root.left, level + 1)
  if root.left == None and root.right == None and level % 2 == 0:
    result += root.elem
  return result
```
