```python
import numpy as np
```

### **Node Class**

```python
class Node:
    def __init__(self, key, value, next):
        self.key = key
        self.value = value
        self.next = next
```

### **HashTable Class**

```python
class HashTable:

    def __init__(self, size):
      self.hashtable = np.array([None]*size)

    def insert(self, key, value):
      index = self.hashFunction(key)
      newNode = Node(key, value, None)
      if self.hashtable[index] == None:
        self.hashtable[index] = newNode
      else:
        newNode.next = self.hashtable[index]
        self.hashtable[index] = newNode

    def search(self, key):
      index = self.hashFunction(key)
      temp = self.hashtable[index]
      while temp != None:
        if temp.key == key:
          return temp.value
        temp = temp.next
      return None

    def hashFunction(self, key):
      return (key * 100) % len(self.hashtable)

    def printHashTable(self):
        for i in range(len(self.hashtable)):
            print(f"Index: {i}")
            temp = self.hashtable[i]
            while temp != None:
              print((temp.key, temp.value), end = "-->")
              temp = temp.next
            print()
```

```python
ht = HashTable(5)
ht.insert(12, "Apple")
ht.insert(5, "Orange")
ht.insert(17, "Banana")
ht.insert(10, "Grapes")
ht.insert(22, "Watermelon")
ht.printHashTable()
value = ht.search(12)
print(value)
```
