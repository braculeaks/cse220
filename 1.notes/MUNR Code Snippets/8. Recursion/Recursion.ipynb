Method Jumping

```python
def methodA(x):
  print("Method A: ", x)
  x = methodB(x + 1)
  print("Method A: ", x)

def methodB(x):
  print("Method B: ", x)
  return x + 2
```

```python
methodA(2)
```

Infinite Recursion

```python
def methodC(x):
  print("Method C: ", x)
  methodC(x + 1)
```

```python
methodC(2)
```

Recursion

```python
def methodD(x):
  if x == 15: #Base Case
    return
  else: #Recursive Case
    print("Method D: ", x)
    methodD(x + 1)
```

```python
methodD(2)
```

Finding Factorial (Iterative)

```python
def factorialIterative(x):
  product = 1
  for i in range(1, x + 1):
    product = product * i
  print(product)
```

```python
factorialIterative(5)
```

Finding Factorial (Recursive)

```python
def factorial(x):
  if x == 1:
    return 1
  else:
    return x * factorial(x - 1)
```

```python
factorial(5)
```

Finding Exponent (Recursive)

```python
def exponent(x, p):
  if p == 0:
    return 1
  else:
    return x * exponent(x, p - 1)
```

```python
exponent(5, 3)
```
