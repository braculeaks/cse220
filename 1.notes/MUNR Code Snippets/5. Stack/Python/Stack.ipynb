```python
class Node:

  def __init__(self, elem, next = None):
    self.elem = elem
    self.next = next
```

```python
class Stack:

  def __init__(self):
    self.top = None

  def push(self, elem):
    newNode = Node(elem, None)
    newNode.next = self.top
    self.top = newNode

  def pop(self):
    if self.top == None:
      return -1
    else:
      removed = self.top
      self.top = self.top.next
      return removed.elem

  def peek(self):
    if self.top == None:
      return -1
    else:
      return self.top.elem

  def isEmpty(self):
    if self.top == None:
      return True
    else:
      return False
```

```python
st = Stack()
st.push(4) #Pushes 4 into stack
st.push(3) #Pushes 3 into stack
st.push(5) #Pushes 5 into stack
st.push(1) #Pushes 1 into stack
st.push(9) #Pushes 9 into stack

print('Peeked Element: ',st.peek()) #Should print 9
print('Popped Element: ',st.pop())  #Should print 9
print('Popped Element: ',st.pop())  #Should print 1
print('Popped Element: ',st.pop())  #Should print 5
print('Peeked Element: ',st.peek()) #Should print 3
print('Popped Element: ',st.pop())  #Should print 3
print('Popped Element: ',st.pop())  #Should print 4
print('Peeked Element: ',st.peek()) #Should print -1
print('Popped Element: ',st.pop())  #Should print -1
print(st.isEmpty())
```
