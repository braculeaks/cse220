```python
class Node:

  def __init__(self, elem, next):
    self.elem = elem
    self.next = next
```

```python
class Queue:

  def __init__(self):
     self.front = None
     self.rear = None

  def enqueue(self, elem):
    newNode = Node(elem, None)
    if self.front == None:
      self.front = newNode
      self.rear = newNode
    else:
      self.rear.next = newNode
      self.rear = newNode

  def dequeue(self):
    if self.front == None:
      return -1
    else:
      removed = self.front
      self.front = self.front.next
      return removed.elem

  def peek(self):
    if self.front == None:
      return -1
    else:
      return self.front.elem

  def isEmpty(self):
    if self.front == None:
      return True
    else:
      return False
```

```python
q = Queue() #Creating an empty queue object
q.enqueue(10) #Inserting 10 into Queue
q.enqueue(11) #Inserting 11 into Queue
q.enqueue(12) #Inserting 12 into Queue
q.enqueue(13) #Inserting 13 into Queue
q.enqueue(14) #Inserting 14 into Queue
print(f"Dequeued element: {q.dequeue()}") #Removes 10 from Queue
print(f"Dequeued element: {q.dequeue()}") #Removes 11 from Queue
print(f"Peeked element: {q.peek()}") #Returns 12 from Queue
print(q.isEmpty()) #Returns False
print(f"Dequeued element: {q.dequeue()}") #Removes 12 from Queue
print(f"Dequeued element: {q.dequeue()}") #Removes 13 from Queue
print(f"Peeked element: {q.peek()}") #Returns 14 from Queue
print(f"Dequeued element: {q.dequeue()}") #Removes 14 from Queue
print(f"Dequeued element: {q.dequeue()}") #Returns -1 (Queue Underflow)
print(f"Peeked element: {q.peek()}") #Returns -1 (Queue Underflow)
print(q.isEmpty())
```
