```python
class BSTNode:

    def __init__(self, elem):
        self.elem = elem
        self.left = None
        self.right = None
```

```python
def inorder(node):
    if node == None:
        return
    inorder(node.left)
    print(node.elem, end = " ")
    inorder(node.right)
```

```python
def insert(root, value):
  if root == None:
    n = BSTNode(value)
    return n
  if value < root.elem:
    root.left = insert(root.left, value)
  elif value > root.elem:
    root.right = insert(root.right, value)
  else:
    print("Duplicate values are not allowed.")
  return root
```

```python
arr = [70, 50, 40, 90, 20, 60, 20, 95, 99, 80, 85, 75]
root = None
for i in range(len(arr)):
  print(f'Inserting value {arr[i]}..')
  root = insert(root, arr[i])
```

```python
inorder(root)
```

```python
def delete(root, value):
  if root == None:
    return root
  if value < root.elem:
    root.left = delete(root.left, value)
    return root
  elif value > root.elem:
    root.right = delete(root.right, value)
    return root
  else:
    if root.left == None and root.right == None: #Case 1
      return None
    elif root.left != None and root.right == None: #Case 2
      return root.left
    elif root.left == None and root.right != None: #Case 2
      return root.right
    else: #Case 3
      successorNode = successor(root.right) #You can delete by predecessor as well. Do it yourself.
      root.elem = successorNode.elem
      root.right = delete(root.right, successorNode.elem)
      return root

def successor(node): #Function to find successor
  while node.left != None:
    node = node.left
  return node
```
