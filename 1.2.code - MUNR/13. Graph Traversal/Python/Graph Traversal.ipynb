```python
import numpy as np
```

```python
n = 8
```

```python
adjMat = np.zeros((n + 1, n + 1), dtype = int)
```

```python
def addEdgeToMatrix(source, destination):
  adjMat[source][destination] = 1
  adjMat[destination][source] = 1
```

```python
addEdgeToMatrix(1, 2)
addEdgeToMatrix(1, 4)
addEdgeToMatrix(2, 3)
addEdgeToMatrix(2, 5)
addEdgeToMatrix(3, 4)
addEdgeToMatrix(3, 6)
addEdgeToMatrix(3, 8)
addEdgeToMatrix(6, 7)
```

```python
print(adjMat)
```

```python
from queue import Queue
```

```python
def bfs(adjMat, source):
  visited = np.zeros(n + 1, dtype = int)
  q = Queue()
  visited[source] = 1
  q.put(source)
  while q.empty() != True:
    u = q.get()
    print(u, end = " ")
    for v in range(len(adjMat[u])):
      if adjMat[u][v] == 1 and visited[v] == 0:
        visited[v] = 1
        q.put(v)
```

```python
bfs(adjMat, 1)
```

```python
from queue import LifoQueue
```

```python
def dfs(adjMat, source):
  visited = np.zeros(n + 1, dtype = int)
  stack = LifoQueue()
  visited[source] = 1
  stack.put(source)
  while stack.empty() != True:
    u = stack.get()
    print(u, end = " ")
    for v in range(len(adjMat[u])):
      if adjMat[u][v] == 1 and visited[v] == 0:
        visited[v] = 1
        stack.put(v)
```

```python
dfs(adjMat, 1)
```
