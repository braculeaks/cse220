```python
# class EdgeNode:
#     def __init__(self, toV, weight):
#         self.toV = toV
#         self.weight = weight
#         self.next = None

#OR
class EdgeNode:
    def __init__(self, fromV, toV, weight):
        self.fromV = fromV
        self.toV = toV
        self.weight = weight
        self.next = None
```

Helper Methods for Adjacency List Representation

```python
import numpy as np

def showAdjList(adj_List):
    for i in range(len(adj_List)):
        print(f"Vertex {i}:", end="")
        printLL(adj_List[i])

def printLL(head):
    n = head
    while n is not None:
        print(f" -> ({n.toV},{n.weight})", end="")
        n = n.next
    print()

def appendLL(head, eNode):
    n = head
    while n.next is not None:
        n = n.next
    n.next = eNode

def addEdge(adj_List, frm, to, weight, undirected=True):
    # Add forward edge
    newNode = EdgeNode(to, weight)
    if adj_List[frm] is None:
        adj_List[frm] = newNode
    else:
        appendLL(adj_List[frm], newNode)

    # Add backward edge if undirected
    if undirected:
        revNode = EdgeNode(frm, weight)
        if adj_List[to] is None:
            adj_List[to] = revNode
        else:
            appendLL(adj_List[to], revNode)

# === Tester Code ===

V = 5  # number of vertices
adj_List = np.array([None] * V, dtype=EdgeNode)

"""
    Example Graph:
         0 --(2)-- 1
        |   \
        (3)  (6)
        |        \
        2 --(1)-- 3
        |
        (4)
        |
        4
"""

addEdge(adj_List, 0, 1, 2)
addEdge(adj_List, 0, 2, 3)
addEdge(adj_List, 0, 3, 6)
addEdge(adj_List, 2, 3, 1)
addEdge(adj_List, 2, 4, 4)

print("Adjacency List Representation:")
print("Vertex [from]: -> (to,weight) -> (to,weight) -> (to,weight)")
showAdjList(adj_List)
```
